```python
import numpy as np

print("=== Method 1: Matrix Addition using Python Lists ===")

matrix_a = [
    [1, 2],
    [3, 4]
]

matrix_b = [
    [5, 6],
    [7, 8]
]

result_list = [
    [0, 0],
    [0, 0]
]

for i in range(len(matrix_a)):
    for j in range(len(matrix_a[0])):
        result_list[i][j] = matrix_a[i][j] + matrix_b[i][j]

print("Matrix A:")
for row in matrix_a:
    print(row)

print("\nMatrix B:")
for row in matrix_b:
    print(row)

print("\nSum (List Addition):")
for row in result_list:
    print(row)

print("\n=== Method 2: Matrix Addition using NumPy Arrays ===")

np_a = np.array([
    [1, 2],
    [3, 4]
])

np_b = np.array([
    [5, 6],
    [7, 8]
])

result_numpy = np_a + np_b

print("NumPy Array A:\n", np_a)
print("\nNumPy Array B:\n", np_b)
print("\nSum (NumPy Addition):\n", result_numpy)
```
